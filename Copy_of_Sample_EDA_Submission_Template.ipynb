## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
The mean hour of occurrence is the same across all crime types.
(i.e., crime type has no effect on the average time crimes occur.)

Alternative Hypothesis (H₁):
The mean hour of occurrence differs between at least one pair of crime types.
(i.e., some crime types tend to occur at different times of the day.)

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
import pandas as pd
import scipy.stats as stats

# Load train dataset
train = pd.read_excel("Train.xlsx", engine="openpyxl")

# Filter relevant columns
data = train[['TYPE', 'HOUR']].dropna()

# Prepare groups
groups = [data[data['TYPE'] == crime]['HOUR'] for crime in data['TYPE'].unique()]

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

print("One-way ANOVA Results")
print("F-statistic:", f_stat)
print("p-value:", p_value)

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis (H₀): Crime types occur at significantly different times of the day.")
else:
    print("Fail to Reject Null Hypothesis (H₀): No significant difference in mean crime times across types.")
````

##### Which statistical test have you done to obtain P-Value?

Since we are comparing mean HOUR (numeric) across multiple crime types (categorical), the appropriate test is:

One-Way ANOVA (Analysis of Variance).

##### Why did you choose the specific statistical test?

Dependent variable: HOUR (numeric, continuous: 0–23).

Independent variable: TYPE (categorical, multiple groups).

We want to test if group means differ.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
The distribution of crime types is independent of neighborhoods.
(i.e., neighborhood and crime type are not related.)

Alternative Hypothesis (H₁):
The distribution of crime types depends on neighborhoods.
(i.e., some neighborhoods have different crime patterns.)

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
import pandas as pd
import scipy.stats as stats

# Load train dataset
train = pd.read_excel("Train.xlsx", engine="openpyxl")

# Contingency table (TYPE vs NEIGHBOURHOOD)
contingency_table = pd.crosstab(train['NEIGHBOURHOOD'], train['TYPE'])

# Chi-Square Test of Independence
chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print("Chi-Square Test Results")
print("Chi2 Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis (H₀): Crime distribution depends on neighborhoods.")
else:
    print("Fail to Reject Null Hypothesis (H₀): Crime distribution is independent of neighborhoods.")
````

##### Which statistical test have you done to obtain P-Value?

Since both variables are categorical (TYPE and NEIGHBOURHOOD), we use the Chi-Square Test of Independence.

##### Why did you choose the specific statistical test?

It checks whether two categorical variables are independent.

Suitable for contingency tables (crime type × neighborhood).

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
The mean number of crimes per month is the same across all years.
(i.e., no trend in monthly crime counts across years.)

Alternative Hypothesis (H₁):
The mean number of crimes per month differs across years.
(i.e., crime frequency changes year-to-year.)

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Group data: monthly counts per year
monthly_counts = train.groupby(['YEAR', 'MONTH']).size().reset_index(name='Incident_Counts')

# Prepare groups for ANOVA
groups = [monthly_counts[monthly_counts['YEAR'] == year]['Incident_Counts']
          for year in monthly_counts['YEAR'].unique()]

# Perform one-way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

print("One-way ANOVA Results (Monthly Counts across Years)")
print("F-statistic:", f_stat)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject Null Hypothesis (H₀): Monthly crime frequency differs across years.")
else:
    print("Fail to Reject Null Hypothesis (H₀): No significant difference in monthly crime frequency across years.")
````

##### Which statistical test have you done to obtain P-Value?

Since we compare monthly crime counts (numeric) across multiple years (categorical), we again use One-Way ANOVA.

Dependent variable: monthly incident counts (continuous).

## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
# Check missing values
train.isnull().sum()

# Drop rows with all NaNs (if any)
train.dropna(how='all', inplace=True)

# Impute numerical columns (e.g., X, Y, Latitude, Longitude, HOUR, MINUTE)
from sklearn.impute import SimpleImputer

num_cols = train.select_dtypes(include=['float64','int64']).columns
imputer_num = SimpleImputer(strategy='median')
train[num_cols] = imputer_num.fit_transform(train[num_cols])

# Impute categorical columns (e.g., TYPE, NEIGHBOURHOOD, HUNDRED_BLOCK)
cat_cols = train.select_dtypes(include=['object']).columns
imputer_cat = SimpleImputer(strategy='most_frequent')
train[cat_cols] = imputer_cat.fit_transform(train[cat_cols])
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Median imputation for numeric (robust against outliers).

Most frequent imputation for categorical (keeps most common category).

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
import numpy as np

# Example: treating outliers in numeric features (Latitude, Longitude, Hour, Minute)
def cap_outliers(df, col, lower=0.01, upper=0.99):
    lower_bound = df[col].quantile(lower)
    upper_bound = df[col].quantile(upper)
    df[col] = np.clip(df[col], lower_bound, upper_bound)

for col in ['X','Y','Latitude','Longitude','HOUR','MINUTE']:
    cap_outliers(train, col)
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

````python
import re

def clean_text(text):
    text = re.sub(r'http\S+|www\S+', '', text)   # Remove URLs
    text = re.sub(r'\w*\d\w*', '', text)         # Remove words with digits
    return text

train['HUNDRED_BLOCK'] = train['HUNDRED_BLOCK'].apply(clean_text)
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    tokens = text.split()
    tokens = [word for word in tokens if word not in stop_words]
    return " ".join(tokens)

train['HUNDRED_BLOCK'] = train['HUNDRED_BLOCK'].apply(remove_stopwords)

# Remove extra whitespaces
train['HUNDRED_BLOCK'] = train['HUNDRED_BLOCK'].str.strip()
````

#### 6. Rephrase Text

````python
# Rephrase Text
from nltk.corpus import wordnet

def synonym_replacement(text):
    words = text.split()
    new_words = []
    for word in words:
        synonyms = wordnet.synsets(word)
        if synonyms:
            new_words.append(synonyms[0].lemmas()[0].name())  # Pick first synonym
        else:
            new_words.append(word)
    return " ".join(new_words)

train['HUNDRED_BLOCK'] = train['HUNDRED_BLOCK'].apply(synonym_replacement)
````

````python
import pandas as pd
import numpy as np
from scipy.stats.mstats import winsorize
from sklearn.preprocessing import LabelEncoder

# -----------------------------
# 1️⃣ Handle Missing Values
# -----------------------------

# Numeric columns → fill with median
numeric_cols = train.select_dtypes(include=['int64', 'float64']).columns
for col in numeric_cols:
    train[col] = train[col].fillna(train[col].median())

# Categorical columns → fill with mode
cat_cols = train.select_dtypes(include=['object']).columns
for col in cat_cols:
    # Convert list entries to comma-separated string
    if train[col].apply(lambda x: isinstance(x, list)).any():
        train[col] = train[col].apply(lambda x: ','.join(x) if isinstance(x, list) else x)

    # Fill missing values
    mode_val = train[col].mode()
    if not mode_val.empty:
        train[col] = train[col].fillna(mode_val.iloc[0])
    else:
        train[col] = train[col].fillna('missing')

# -----------------------------
# 2️⃣ Outlier Treatment
# -----------------------------

for col in numeric_cols:
    train[col] = winsorize(train[col], limits=[0.01, 0.01])  # clip 1% lowest/highest values

# -----------------------------
# 3️⃣ Encode Categorical Features
# -----------------------------

le = LabelEncoder()
for col in cat_cols:
    train[col] = le.fit_transform(train[col])

# -----------------------------
# 4️⃣ Create New Features (Optional)
# -----------------------------

# Example: Word count from text column
if 'HUNDRED_BLOCK' in train.columns:
    train['HUNDRED_BLOCK_word_count'] = train['HUNDRED_BLOCK'].astype(str).apply(lambda x: len(x.split()))

# You can add more feature manipulations here as needed

# -----------------------------
# ✅ Final Check
# -----------------------------
print("Dataset ready after feature manipulation:")
print(train.head())
print("\nNumeric columns:", numeric_cols)
print("Categorical columns:", cat_cols)
````
